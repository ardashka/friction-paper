```python
from CID import InterlacedTime
import numpy as np
import time
```

```python
p = 5
it = InterlacedTime(p)
data = np.random.randint(2, size=(4<<p, 1<<p, 1<<p))

start = time.time()
res = it.isotropic_q_order(data)
end = time.time()
print('time :', end - start)
print(res)
```

```python
data = np.append(np.zeros((2<<p, 1<<p, 1<<p)), np.ones((2<<p, 1<<p, 1<<p)), axis=0)
print(data.shape)
start = time.time()
res = it.isotropic_q_order(data)
end = time.time()
print('time :', end - start)
print(res)
```
