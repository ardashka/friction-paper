```python
from HilbertCurve import HilbertCurve
from matplotlib import pyplot as plt
# import matplotlib.pyplot as plt
import numpy as np
```

```python
def hilbert2D(data):
    x0 = [x[0] for x in data]
    x1 = [x[1] for x in data]
    plt.figure(figsize=(10,10))
    plt.plot(x0,x1)
    plt.gca().invert_yaxis()
    plt.show()
```

```python
hilbert = HilbertCurve(5, 2)
points = hilbert.points
hilbert2D(points)
```
