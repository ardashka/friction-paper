```python
from HilbertCurve import hilbert_principle_curve, hilbert_3D_curves
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import pyplot as plt
import time
```

```python
def hilbert2D(points):
    x0 = [point[0] for point in points]
    x1 = [point[1] for point in points]
    plt.figure(figsize=(6,6))
    plt.plot(x0,x1)
    plt.xlabel('time')
    plt.ylabel('x')
    plt.gca().xaxis.tick_top()
    plt.gca().invert_yaxis()
    plt.show()
    
def hilbert3D(points):
    x0 = [point[0] for point in points]
    x1 = [point[1] for point in points]
    x2 = [point[2] for point in points]
    plt.figure(figsize=(6,6))
    ax = plt.axes(projection="3d")
    ax.plot(x0,x1,x2)
    ax.set_xlabel('time')
    ax.set_ylabel('x')
    ax.set_zlabel('y')
    plt.gca().xaxis.tick_top()
    plt.gca().invert_yaxis()
    plt.show()
```

```python
hilbert2D(hilbert_principle_curve(5, 2))
```

```python
for hilbert_curve in hilbert_3D_curves(1):
    hilbert3D(hilbert_curve)
```
